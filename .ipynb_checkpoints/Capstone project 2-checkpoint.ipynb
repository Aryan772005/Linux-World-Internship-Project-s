```python
features_2d = ['Annual Income (k$)', 'Spending Score (1-100)']
X_2d = df[features_2d]
scaler_2d = StandardScaler()
X_2d_scaled = scaler_2d.fit_transform(X_2d)
```

```python
wcss = []
K_range = range(1, 11)
for k in K_range:
    km = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init=10)
    km.fit(X_2d_scaled)
    wcss.append(km.inertia_)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(K_range, wcss, marker='o', linestyle='--', color='b', markersize=8)
plt.title('Elbow Method - Finding Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS (Within-Cluster Sum of Squares)')
plt.xticks(K_range)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
sil_scores = []
for k in range(2, 11):
    km = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init=10)
    labels = km.fit_predict(X_2d_scaled)
    score = silhouette_score(X_2d_scaled, labels)
    sil_scores.append(score)
    print(f'K={k}, Silhouette Score={score:.4f}')
```

```python
plt.figure(figsize=(10, 6))
plt.plot(range(2, 11), sil_scores, marker='s', linestyle='--', color='red', markersize=8)
plt.title('Silhouette Score for Different K Values')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.xticks(range(2, 11))
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
optimal_k = 5
kmeans_final = KMeans(n_clusters=optimal_k, init='k-means++', random_state=42, n_init=10)
df['Cluster'] = kmeans_final.fit_predict(X_2d_scaled)
```

```python
df['Cluster'].value_counts().sort_index()
```

```python
centers = scaler_2d.inverse_transform(kmeans_final.cluster_centers_)
plt.figure(figsize=(12, 8))
scatter = sns.scatterplot(data=df, x='Annual Income (k$)', y='Spending Score (1-100)', hue='Cluster', palette='Set1', s=100, alpha=0.7, edgecolor='black', linewidth=0.5)
plt.scatter(centers[:, 0], centers[:, 1], c='black', marker='X', s=300, label='Centroids')
plt.title('Customer Segments (K-Means Clustering, K=5)')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
cluster_profile = df.groupby('Cluster').agg(
    Count=('CustomerID', 'count'),
    Avg_Age=('Age', 'mean'),
    Avg_Income=('Annual Income (k$)', 'mean'),
    Avg_Spending=('Spending Score (1-100)', 'mean'),
    Min_Income=('Annual Income (k$)', 'min'),
    Max_Income=('Annual Income (k$)', 'max'),
    Min_Spending=('Spending Score (1-100)', 'min'),
    Max_Spending=('Spending Score (1-100)', 'max')
).reset_index()
cluster_profile
```

```python
features_3d = ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']
X_3d = df[features_3d]
scaler_3d = StandardScaler()
X_3d_scaled = scaler_3d.fit_transform(X_3d)
kmeans_3d = KMeans(n_clusters=5, init='k-means++', random_state=42, n_init=10)
df['Cluster_3D'] = kmeans_3d.fit_predict(X_3d_scaled)
```

```python
fig = plt.figure(figsize=(14, 10))
ax = fig.add_subplot(111, projection='3d')
scatter = ax.scatter(df['Age'], df['Annual Income (k$)'], df['Spending Score (1-100)'], c=df['Cluster_3D'], cmap='Set1', s=60, alpha=0.7)
ax.set_xlabel('Age')
ax.set_ylabel('Annual Income (k$)')
ax.set_zlabel('Spending Score')
ax.set_title('3D View of Customer Clusters')
plt.colorbar(scatter, label='Cluster')
plt.tight_layout()
plt.show()
```

```python
linked = linkage(X_2d_scaled, method='ward')
plt.figure(figsize=(16, 8))
dendrogram(linked, truncate_mode='lastp', p=30, leaf_rotation=90, leaf_font_size=10, show_contracted=True)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Cluster Size')
plt.ylabel('Distance')
plt.tight_layout()
plt.show()
```

```python
agg_clustering = AgglomerativeClustering(n_clusters=5, linkage='ward')
df['Cluster_Hierarchical'] = agg_clustering.fit_predict(X_2d_scaled)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(18, 7))
sns.scatterplot(data=df, x='Annual Income (k$)', y='Spending Score (1-100)', hue='Cluster', palette='Set1', s=80, ax=axes[0], edgecolor='black', linewidth=0.3)
axes[0].set_title('K-Means Clustering')
sns.scatterplot(data=df, x='Annual Income (k$)', y='Spending Score (1-100)', hue='Cluster_Hierarchical', palette='Set1', s=80, ax=axes[1], edgecolor='black', linewidth=0.3)
axes[1].set_title('Hierarchical Clustering')
plt.tight_layout()
plt.show()
```

```python
km_sil = silhouette_score(X_2d_scaled, df['Cluster'])
hc_sil = silhouette_score(X_2d_scaled, df['Cluster_Hierarchical'])
comparison = pd.DataFrame({
    'Algorithm': ['K-Means', 'Hierarchical'],
    'Silhouette Score': [km_sil, hc_sil]
})
comparison
```

```python
cluster_names = {
    0: 'Low Income, Low Spending',
    1: 'High Income, High Spending',
    2: 'Average Income, Average Spending',
    3: 'High Income, Low Spending',
    4: 'Low Income, High Spending'
}
df['Segment_Name'] = df['Cluster'].map(cluster_names)
df[['CustomerID', 'Gender', 'Age', 'Annual Income (k$)', 'Spending Score (1-100)', 'Cluster', 'Segment_Name']].head(10)
```

```python
segment_summary = df.groupby('Segment_Name').agg(
    Total_Customers=('CustomerID', 'count'),
    Avg_Age=('Age', 'mean'),
    Avg_Income=('Annual Income (k$)', 'mean'),
    Avg_Spending=('Spending Score (1-100)', 'mean')
).round(2)
segment_summary
```

```python
plt.figure(figsize=(12, 6))
segment_summary['Total_Customers'].plot(kind='barh', color=['#e74c3c', '#2ecc71', '#3498db', '#f39c12', '#9b59b6'])
plt.title('Number of Customers per Segment')
plt.xlabel('Count')
plt.ylabel('Segment')
plt.tight_layout()
plt.show()
```

```python
def assign_strategy(row):
    segment = row['Segment_Name']
    if segment == 'High Income, High Spending':
        return 'VIP/Premium Loyalty Program'
    elif segment == 'High Income, Low Spending':
        return 'Targeted Re-engagement Campaign'
    elif segment == 'Low Income, High Spending':
        return 'Upsell with Affordable Premium Options'
    elif segment == 'Average Income, Average Spending':
        return 'Retention via Personalized Offers'
    elif segment == 'Low Income, Low Spending':
        return 'Budget-Friendly Deals & Discounts'
    else:
        return 'General Marketing'

df['Targeting_Strategy'] = df.apply(assign_strategy, axis=1)
```

```python
df[['CustomerID', 'Gender', 'Age', 'Annual Income (k$)', 'Spending Score (1-100)', 'Segment_Name', 'Targeting_Strategy']].head(15)
```

```python
strategy_counts = df['Targeting_Strategy'].value_counts()
plt.figure(figsize=(12, 6))
colors = ['#1abc9c', '#e74c3c', '#3498db', '#f1c40f', '#9b59b6']
strategy_counts.plot(kind='bar', color=colors, edgecolor='black')
plt.title('Distribution of Targeting Strategies')
plt.xlabel('Targeting Strategy')
plt.ylabel('Number of Customers')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

```python
strategy_cluster = pd.crosstab(df['Targeting_Strategy'], df['Cluster'])
strategy_cluster
```
