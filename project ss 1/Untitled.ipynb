# Business Problem Framing

## Business Problem

Hotels often lose revenue because of booking cancellations and inefficient room planning. The management wants to understand customer booking behavior, identify factors affecting cancellations, and improve operational decisions.

## Data Science Problem Statement

Analyze historical hotel booking data to discover booking patterns, identify factors influencing cancellations, and generate insights that can support better business decisions. The dataset can also be used to build a machine learning model to predict whether a booking is likely to be canceled.

## Business Objective

• Reduce booking cancellations
• Improve room occupancy
• Increase hotel revenue
• Understand customer booking behavior
• Support better planning and decision-making

## Required Data

- Hotel Type
- Lead Time
- Arrival Date
- Number of Adults
- Number of Children
- Number of Babies
- Meal
- Market Segment
- Distribution Channel
- Country
- Reserved Room Type
- Assigned Room Type
- Booking Changes
- Deposit Type
- Customer Type
- Average Daily Rate (ADR)
- Special Requests
- Booking Cancellation Status

## Expected Analytical Output

- Cancellation trends
- Seasonal booking patterns
- Hotel-wise booking comparison
- Customer type analysis
- Country-wise booking distribution
- Average Daily Rate analysis
- Effect of lead time on cancellations

## Expected Machine Learning Output

Predict whether a hotel booking will be canceled based on customer and booking information.

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df = df.drop("company", axis=1)
```

```python
missing = (df.isnull().sum() / len(df)) * 100
missing[missing > 0]
```

```python
df["children"] = df["children"].fillna(0)
```

```python
df["country"] = df["country"].fillna(df["country"].mode()[0])
```

```python
df["agent"] = df["agent"].fillna(0)
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df["reservation_status_date"] = pd.to_datetime(df["reservation_status_date"])
```

```python
df.info()
```

```python
df.to_csv("hotel_bookings_cleaned.csv", index=False)
```

### Data Cleaning Summary

- Removed duplicate records.
- Dropped the company column because it contained a large number of missing values.
- Filled missing values in the children column with 0.
- Filled missing values in the country column using the most frequent country.
- Filled missing values in the agent column with 0.
- Converted the reservation status date to datetime format.
- Saved the cleaned dataset for further analysis.

```python
plt.figure(figsize=(6,4))

sns.countplot(x="is_canceled", data=df)

plt.title("Booking Cancellation Distribution")
plt.xlabel("Cancellation Status")
plt.ylabel("Number of Bookings")

plt.show()
```

```python
month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

plt.figure(figsize=(10,5))

sns.countplot(
    x="arrival_date_month",
    data=df,
    order=month_order
)

plt.xticks(rotation=45)

plt.title("Monthly Booking Trend")
plt.xlabel("Month")
plt.ylabel("Bookings")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["lead_time"], bins=40)

plt.title("Lead Time Distribution")
plt.xlabel("Lead Time")
plt.ylabel("Frequency")

plt.show()
```
