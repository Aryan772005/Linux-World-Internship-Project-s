```python
cluster_names = {
    0: 'Low Income, Low Spending',
    1: 'High Income, High Spending',
    2: 'Average Income, Average Spending',
    3: 'High Income, Low Spending',
    4: 'Low Income, High Spending'
}
df['Segment_Name'] = df['Cluster'].map(cluster_names)
df[['CustomerID', 'Gender', 'Age', 'Annual Income (k$)', 'Spending Score (1-100)', 'Cluster', 'Segment_Name']].head(10)
```

```python
segment_summary = df.groupby('Segment_Name').agg(
    Total_Customers=('CustomerID', 'count'),
    Avg_Age=('Age', 'mean'),
    Avg_Income=('Annual Income (k$)', 'mean'),
    Avg_Spending=('Spending Score (1-100)', 'mean')
).round(2)
segment_summary
```

```python
plt.figure(figsize=(12, 6))
segment_summary['Total_Customers'].plot(kind='barh', color=['#e74c3c', '#2ecc71', '#3498db', '#f39c12', '#9b59b6'])
plt.title('Number of Customers per Segment')
plt.xlabel('Count')
plt.ylabel('Segment')
plt.tight_layout()
plt.show()
```

```python
def assign_strategy(row):
    segment = row['Segment_Name']
    if segment == 'High Income, High Spending':
        return 'VIP/Premium Loyalty Program'
    elif segment == 'High Income, Low Spending':
        return 'Targeted Re-engagement Campaign'
    elif segment == 'Low Income, High Spending':
        return 'Upsell with Affordable Premium Options'
    elif segment == 'Average Income, Average Spending':
        return 'Retention via Personalized Offers'
    elif segment == 'Low Income, Low Spending':
        return 'Budget-Friendly Deals & Discounts'
    else:
        return 'General Marketing'

df['Targeting_Strategy'] = df.apply(assign_strategy, axis=1)
```
